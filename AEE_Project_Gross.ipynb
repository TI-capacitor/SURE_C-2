```python
import matplotlib.pyplot as plt
import statistics
import pandas as pd
import datetime
import numpy as np
import scipy.stats as st
# parse_dates: especificar cual columna contiene fechas para leerlas correctamente
# Date_format: especificar cual es el formato de las fechas en el archive
# encoding: especificar el encoding de el archivo que se va a leer. Por tener acentos se utiliza latin-1
# thousands: utilisado para especifiar que los miles estan siendo separados por una coma


data = pd.read_csv('AEEGENERACION.csv',
                   parse_dates=['Mes'],
                   date_format="%m/%d/%Y",
                   encoding="latin-1",thousands=',')
#since the date was in D/M/Y, it needed to be changed to M/D/Y

dataBefore2017 = data.drop(data
    [data.Mes > np.datetime64
         (datetime.date(2017, 1, 1))].index)
# Se eliminan todos los datos despues de 2017

dataAfter2017 = data.drop(data
     [data.Mes < np.datetime64
        (datetime.date(2017, 1, 1))].index) 
# Se eliminan todos los datos antes de 2017

stdevBefore2017= np.std(dataBefore2017["Gross Generation"])
print(stdevBefore2017)

stdevAfter2017 = np.std(dataAfter2017["Gross Generation"])
print(stdevAfter2017)
meanBefore2017 = dataBefore2017["Gross Generation"].mean() 
meanAfter2017 = dataAfter2017["Gross Generation"].mean()
#Average of gross generation is calculated for both
#before and after the natural disasters

varBefore2017 = dataBefore2017["Gross Generation"].var()
varAfter2017 = dataAfter2017["Gross Generation"].var()
mean1 = data["Gross Generation"].mean()

t_stat, pvalue = st.ttest_ind(
    dataBefore2017["Gross Generation"],
        dataAfter2017["Gross Generation"], equal_var=True, alternative="greater" )

print(t_stat, pvalue)
print("Mean pre 2017: " + str(meanBefore2017))
print("Mean post 2017: " + str(meanAfter2017))

print("Var pre 2017: " + str(varBefore2017))
print("Var post 2017: " + str(varAfter2017))



plt.plot(data["Mes"], data["Gross Generation"], 'green') #Dibujar la grafica con valores x, y
plt.title("Gross generation by date") # textbox for graph title
plt.xlabel("Date")  # Name of abscissa
plt.ylabel("Gross Generation (Mkwh)") # Name of ordenate
plt.show() # Mostrar ventana con grafica dibujada
```
