```python
import matplotlib.pyplot as plt
import pandas as pd
import datetime
import numpy as np

# parse_dates: especificar cual columna contiene fechas para leerlas correctamente
# Date_format: especificar cual es el formato de las fechas en el archive
# encoding: especificar el encoding de el archivo que se va a leer. Por tener acentos se utiliza latin-1
# thousands: utilisado para especifiar que los miles estan siendo separados por una coma


data = pd.read_csv('AEEGENERACION.csv', parse_dates=['Mes'], date_format="%m/%d/%Y", encoding="latin-1",thousands=',')


dataAntes2017 = data.drop(data[data.Mes > np.datetime64(datetime.date(2017, 1, 1))].index) # Se eliminan todos los datos despues de 2017
dataDespues2017 = data.drop(data[data.Mes < np.datetime64(datetime.date(2017, 1, 1))].index) # Se eliminan todos los datos antes de 2017

meanAntes2017 = dataAntes2017["Maximum demand (MW)"].mean()
meanDespues2017 = dataDespues2017["Maximum demand (MW)"].mean()

mean1 = data["Maximum demand (MW)"].mean()

print("Mean pre 2017: " + str(meanAntes2017))
print("Mean post 2017: " + str(meanDespues2017))




plt.plot(data["Mes"], data["Maximum demand (MW)"], 'orange') #Dibujar la grafica con valores x, y
plt.title("Maximum demand by date") # especificar el titulo
plt.xlabel("Date")  # Nombre de el eje de x
plt.ylabel("Maximum demand (MW)") # Nombre de el eje de Y
plt.show() # Mostrar ventana con grafica dibujada
```
